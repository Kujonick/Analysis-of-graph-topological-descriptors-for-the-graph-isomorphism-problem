```python
with open(rf"datasets/raw/brec_400.g6", "w") as merged:
    for table_name in os.listdir(r"datasets/brec/Data/raw"):
        table = np.load(rf"datasets/brec/Data/raw/{table_name}")
        print(table.shape)

        with open(rf"datasets/brec/Data/split/{table_name.split('.')[0]}.g6", "w", encoding="ascii") as f:
            for raw in table.flatten():
                if isinstance(raw, np.str_):
                    raw = str(raw)
                else:
                    raw = raw.decode("ascii")
                f.write(str(raw)+ "\n")
                merged.write(str(raw) + "\n")
```
